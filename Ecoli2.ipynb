```python
import math as mt
def Ecoli (rP, rB, dN, dT):
    #El radio de la tierra en km 6371e3
    rPc = (rP * 1e3) 
    #Raio de E. coli en um 0.5e-6
    rBc = (rB * 1e-6) 
    #Delta T en horas:
    dTc = dT / 60
    #Número de bacterias con respecto a la tierra a) :
    N = (rPc / rBc) ** 3
    N0 = 1
    #Cálculo corregido para horas con la TASA b) :
    thor = dTc * mt.log(N / N0) / mt.log(dN)
    #Para el tiempo que tarda en llenar la tierra en días a) :
    tdys = thor / 24
    print ("Se necesitarían ", N, "bichos para llenar este planeta")
    print (tdys, "días se tardarán los bichos en llenarla")
    #Para calcular la SUPERFICIE b) :
    S = 4 * (rPc / rBc) ** 2
    Shor = dTc * mt.log(S / N0) / mt.log(dN)
    print ("Para la superficie se necesitan: ", S, "bichos")
    print ("Y se tardrían ", Shor, "horas")
    return f"Teniendo las variables: rP ={rP}, rB ={rB}, dN ={dN}, dT ={dT}"
tasa = Ecoli (6371, 0.5, 2, 30)
print (tasa)
```

```python
#Para una tasa de una hora c) i) :
unah = Ecoli (6371, 0.5, 2, 60)
print (unah)
```

```python
#Para una tasa de cuarto de hora c) ii) :
cuarth = Ecoli (6371, 0.5, 2, 15)
print (cuarth)
```
